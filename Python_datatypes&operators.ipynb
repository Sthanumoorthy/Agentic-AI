# **PYTHON OPERATORS**

# Operators are used to perform operations on variables and values.

```python
#Arithmetic Operators
a1=5
a2=2
```

```python
print(a1,a2)
print(a1+a2)
print(a1-a2)
print(a1*a2)
print(a1/a2)
print(a1%a2)
print(a1**a2) #power
print(a1//a2)
```

```python
# Comparison Operators.
print(a1,a2)
print(a1==a2)
print(a1==5)
print(a1!=a2)
print(a1>a2)
print(a1<a2)
print(a1>=a2)
print(a1<=a2)
```

```python
# Logical Operator
#  And gate 1 1=1
# Or gate  =0 0 = 0
#Not gate =0~1,1~0
b1=True
b2=False
```

```python
print(b1,b2)
print(b1 and b2)
print(b1 or b2)
print(not b1)
print(not b2)
```
