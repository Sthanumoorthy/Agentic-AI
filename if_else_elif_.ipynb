# **IF else Statment**

```python
if age >=18:
  print( 'you are eligible to vote' )
else:
  print( 'you are not eligible to vote' )

```

# **If elif else ladder**

```python
#creating a grading system
# 90 =A
#80-90 =B
#70-80=C
# 60-70= D
#50-60=E
```

```python
 mark  =90
 if mark >=90:
   print( 'A' )
 elif mark >=80:
   print( 'B' )
 elif mark >=70:
   print( 'C' )
 elif mark >=60:
   print( 'D' )
```

```python
#grading system
# >=90 -> A ,75 - 90 'B', 60 to 75 'C' ,35 - 60 -> 'D'
# 20 to 35 'E'
# 0 to 20 'F'
score = 85
if score >= 90:
  print('A')
elif score >= 75:
  print('B')
elif score >= 60:
  print('C')
elif score >= 35:
  print('D')
elif score >= 20:
  print('E')
else:
  print('F')
```

```python
#nested condition
age = 25
indian_citizen = True
if age>=18:
  if indian_citizen:
    print('Person is indian citizen and >=18 years, So eligible to vote')
  else:
    print('Person > 18 but not an indian citizen')
else:
  print('Not eligible to vote')
```
