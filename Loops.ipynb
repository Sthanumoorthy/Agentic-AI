# **While Loop** entry control loop

```python
# Print vbalues from 1 to 20
 #initialization,condition,increment/decrement
c =1
while c<=20:
   print(c)
   c =c+1
```

```python


# Print values from 20 to 1
c1 = 20

while c1 >= 1:
    print(c1)
    c1 = c1 - 1

```

## for loop with **range**
range (begining value,end value, step)

```python
#print values from 1-10
for i in range (1,11,1):
  print(i,end=" ")
```

```python
#print values from 1-10
for i in range (1,11,1):
  print(i)
```

```python
# print all the even numbers from 1-20
for i in range(2,21,2):
  print(i)
```

```python
#print from 1-10
for i in range(11):
  print(i)
```

```python
#print from 0-5
for i in range(6):
  print(i)
```

```python
#print numbers from 20-11 in reversew order
for i in range(20,10,-1):
  print(i)
```

# **For Loop with conditional statements**

```python
for i in range(1,11):
      print(i)
```

```python
for i in range(1,11):
  if i%2==0:
    print("even")
  else:
    print("odd")
```

```python
#print form 1- 10 dont print 5
for i in range(1,11):
  if i!=5:
   print(i)
```

```python
#print form 1- 10 dont print 5
for i in range(1,11):
  if i==5:
   pass
  else:
   print(i)
```

```python
#continue is skipping
#print form 1- 10 dont print 5
for i in range(1,11):
  if i==5:
   Continue
  else:
   print(i)
```

```python
#print 5,15,,20,25,30
for i in range(5,35,5):
  if i==10:
    continue
  else:
    print(i)
```

```python
#print
#break is stopping

#print form 1- 10 dont print 5
for i in range(1,11):
  if i==5:
   break
  else:
   print(i)
```

```python
# run a looop from 1-100  stop in 20
for  i in range(1,101):
  if i==21:
    break
  print(i,end =" ")
```
